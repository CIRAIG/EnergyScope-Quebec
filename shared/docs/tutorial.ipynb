# Tutorial on how to use the shared package

In your virtual environment, you can run the following command to install the shared package:
```bash
pip install -e .
```

After that, you can import the package in your Python code and use it to run the model and plot the results, as shown in the code snippets below.

````python
from shared.models import energyscope_original_snapshot_2023
from shared.utils import run_model
from shared.plotting import plot_sankey
from shared.plotting import plot_sankey_co2
from shared.utils import load_snapshot
from shared.utils import run_model
````

````python
results = run_model(energyscope_original_snapshot_2023)
````

````python
plot_sankey(results)
````

````python
plot_sankey_co2(results)
````

````python
snapshot_2050 = load_snapshot(2050)
````

````python
result_2050 = run_model(snapshot_2050)
````

````python
plot_sankey(result_2050)
````

````python
plot_sankey_co2(result_2050)
````
